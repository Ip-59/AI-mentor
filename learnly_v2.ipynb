```python
import IPython
from IPython.display import display, Markdown, HTML
from IPython.core.magic import register_cell_magic
import time
from typing import Callable, Any, Dict

current_task = 1

def show_hint(task_num):
    """Показывает подсказку для указанного задания"""
    if task_num in TASKS:
        display(Markdown(f"💡 **Подсказка:** {TASKS[task_num]['hint']}"))
        create_solution_cell()
    else:
        display(Markdown("❌ Не удалось найти текущее задание"))

def create_solution_cell(title):
    """Создает новую ячейку с магической командой"""
    shell = IPython.get_ipython()
    shell.set_next_input(f'%%check_solution\n\n# {title}\n# Напишите ваше решение здесь:\n\n', replace=False)

def show_task(task_num: int):
    """Показывает задание с указанным номером"""
    if task_num in TASKS:
        display(Markdown(f"### Задание {task_num}\n{TASKS[task_num]['title']}"))
        create_solution_cell(TASKS[task_num]['title'])

# Структура заданий
TASKS = {
    1: {
        'title': 'Создайте переменную x и присвойте ей значение 42',
        'hint': 'Используйте оператор присваивания =, например: variable = value',
        'check': lambda: 'x' in globals() and x == 42,
        'error': 'Нужно создать переменную x со значением 42',
        'success': 'Отлично! Задание выполнено правильно!'
    },
    2: {
        'title': 'Создайте список numbers, содержащий числа от 1 до 5',
        'hint': 'Списки создаются с помощью квадратных скобок: [1, 2, ...]',
        'check': lambda: ('numbers' in globals() and
                         isinstance(numbers, list) and
                         numbers == [1, 2, 3, 4, 5]),
        'error': 'Список numbers должен содержать числа от 1 до 5 в порядке возрастания',
        'success': 'Превосходно! Вы справились со вторым заданием!'
    },
    3: {
        'title': 'Напишите функцию square(n), которая возвращает квадрат числа n',
        'hint': '''Функция определяется так:
def square(n):
    return ...''',
        'check': lambda: ('square' in globals() and
                         callable(square) and
                         all(square(n) == n*n for n in [-2, 0, 4])),
        'error': 'Функция square должна возвращать квадрат числа. Проверьте случаи: square(4)=16, square(0)=0, square(-2)=4',
        'success': 'Отлично! Функция square работает правильно!'
    }
}

# Очищаем пользовательские переменные
to_keep = {
    'IPython', 'display', 'Markdown', 'HTML', 'register_cell_magic',
    'time', 'Callable', 'Any', 'Dict', 'TASKS', 'create_solution_cell',
    'show_hint', 'show_task', 'check_solution'
}

# Очищаем переменные, не затрагивая системные
for var in list(globals().keys()):
    if var not in to_keep and not var.startswith('_'):
        del globals()[var]

# Регистрируем магическую команду для проверки решений
@register_cell_magic
def check_solution(line, cell):
    global current_task
    if 'current_task' not in globals():
        current_task = 1

    # Убираем лишние пробелы и переводы строк
    clean_cell = cell.strip().lower()

    # Проверяем, не является ли это запросом подсказки
    if any(word in clean_cell for word in ['hint', 'help']):
        if current_task in TASKS:
            show_hint(current_task)  # Используем существующую функцию show_hint
        return


    # Выполняем код пользователя в отдельном пространстве имен
    user_globals = globals().copy()
    try:
        exec(cell, user_globals)
        # Обновляем глобальные переменные только после успешного выполнения
        globals().update(user_globals)
    except Exception as e:
        display(Markdown(f"❌ Ошибка в коде: {str(e)}"))
        create_solution_cell()
        return

    # Проверяем текущую задачу и показываем следующую
    if current_task in TASKS:
        task = TASKS[current_task]
        try:
            if task['check']():
                display(Markdown(f"✅ {task['success']}"))
                time.sleep(1)

                # Переходим к следующему заданию
                current_task += 1
                if current_task in TASKS:
                    show_task(current_task)
                else:
                    display(Markdown("""### Поздравляем! 🎉
Вы успешно завершили все задания! Вы научились:
1. Создавать переменные
2. Работать со списками
3. Создавать функции"""))
            else:
                display(Markdown(f"❌ {task['error']}"))
                create_solution_cell()
        except Exception as e:
            display(Markdown(f"❌ Ошибка при проверке: {str(e)}"))
            create_solution_cell()

# Показываем приветствие и инструкции
display(Markdown("""# Добро пожаловать в Learnly! 🎓

Это интерактивный обучающий ноутбук.

📝 Как выполнять задания:
1. В появившейся ячейке напишите своё решение
2. Запустите ячейку (Shift + Enter)
3. Получите обратную связь и следующее задание!

💡 Нужна подсказка?
   Просто напишите слово 'hint' или 'help' в ячейке с %%check_solution
"""))

time.sleep(1)

# Показываем первое задание
current_task = 1
show_task(current_task)
```
