```python
import os
import openai
from dotenv import load_dotenv
from IPython.core.magic import register_cell_magic
from IPython.display import display, Markdown

from learnly.engine import LearnlyEngine


# Загружаем переменные окружения из файла .env
load_dotenv()

# Устанавливаем API ключ для OpenAI
openai.api_key = os.getenv('OPENAI_API_KEY')


# от ChatGPT для Игоря: пример кода (в одной из ячеек learnly.ipynb), который собирает данные от студента
# и вызывает класс LearnlyEngine. Код закомментирован, чтобы пока не выполнялся автоматически.
"""
# Сбор информации о студенте
student_name = input("Введите ваше имя (ФИО): ")
student_email = input("Введите ваш e-mail: ")
hours_planned = int(input("Сколько часов вы планируете потратить на обучение?: "))
difficulty = input("Какой уровень сложности материала предпочитаете? (легкий/средний/сложный): ")
preferred_tasks = input("Какие форматы заданий вам интересны? (multiple_choice, open_question, code_task - через запятую): ")
preferred_tasks_list = [t.strip() for t in preferred_tasks.split(",")]
"""

# от ChatGPT для Игоря: Для теста и отладки, пока просто жестко зададим данные
student_name = "Иванов Иван"
student_email = "ivan@email.ru"
hours_planned = 100
difficulty = "средний"
preferred_tasks_list = ["multiple_choice", "open_question", "code_task"]


# Создаем объект LearnlyEngine - движок для обучения
# engine = LearnlyEngine()
engine = LearnlyEngine(
    student_name=student_name,
    student_email=student_email,
    hours_planned=hours_planned,
    difficulty=difficulty,
    preferred_tasks=preferred_tasks_list
)

@register_cell_magic
def check_solution(line, cell):
   """Магическая команда для проверки решений"""
   try:
       # Выполняем код пользователя
       exec(cell, globals())
       # Проверяем решение
       engine.check_solution(cell, globals())
   except Exception as e:
       display(Markdown(f"❌ Ошибка в коде: {str(e)}"))
       engine.create_solution_cell(engine.tasks[engine.current_task]['problem'])

engine.start_lesson()

# от ChatGPT для Игоря: Далее в ячейках можно работать с engine, запускать основной цикл обучения и т.д.
# Например:
# engine.start_lesson_loop()

print("Инициализация движка завершена. Можно запустить engine.start_lesson_loop() для начала урока.")
```
