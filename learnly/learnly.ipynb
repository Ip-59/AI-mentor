```python
from IPython.core.magic import register_cell_magic
from IPython.display import display, Markdown

from learnly_engine import LearnlyEngine

engine = LearnlyEngine()

@register_cell_magic
def check_solution(line, cell):
   """Магическая команда для проверки решений"""
   try:
       # Выполняем код пользователя
       exec(cell, globals())
       # Проверяем решение
       engine.check_solution(cell, globals())
   except Exception as e:
       display(Markdown(f"❌ Ошибка в коде: {str(e)}"))
       engine.create_solution_cell(engine.tasks[engine.current_task]['title'])

# Показываем приветствие и первое задание
engine.show_welcome()
```
