## Task 1: Implement Linear Regression
Follow the instructions below to complete your implementation of linear regression.

```python
# Import necessary libraries
import numpy as np
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([1, 2, 3, 4, 5])

# Define and train the model
model = LinearRegression().fit(X, y)

# Test the model
print('Model slope:', model.coef_)
print('Model intercept:', model.intercept_)
```

## Task 2: Test Your Model
Now, use the code below to check the accuracy of your model by predicting new values and calculating mean squared error.

```python
# Mean Squared Error Check
from sklearn.metrics import mean_squared_error

# Predict new values
predictions = model.predict(X)

# Calculate mean squared error
mse = mean_squared_error(y, predictions)
print('Mean Squared Error:', mse)

# Simple test to pass assignment
if mse < 1e-5:
    print('Task complete! Well done!')
else:
    print('Please review your model and try again.')
```
